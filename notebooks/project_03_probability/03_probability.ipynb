```python
import pandas as pd
from scipy.stats import norm
```

```python
df = pd.read_csv('../insurance_lab/data/processed/insurance_clean.csv')
df
```

```python
n_smoker = df[df['smoker'] == 'yes'].shape[0]
n_total = len(df)
p_hat = n_smoker / n_total
print(f'n_smoker : {n_smoker} | n_total : {n_total} | p_hat : {p_hat :.4f}')
```

```python
df['smoker'].value_counts(normalize=True)
```

```python
# empirical 
p_empirical = df[df['charges'] > 20000].shape[0]/n_total
print(f'pro of empirical : {p_empirical : .4f}')

pr_empirical = (df['charges'] > 20000).mean()
print(f'prob of empirical : {pr_empirical : .4f}')
```

```python
mean_charges = df['charges'].mean()
std_charges = df['charges'].std()

z_score = (20000 - mean_charges) / std_charges

p_normal = norm.sf(z_score)
print(f'mean :{mean_charges : .2f}')
print(f'std : {std_charges : .2f}')
print(f'z score  : {z_score : .2f}')
print(f'p normal  : {p_normal : .2f}')
```

```python
print(f'p empirical : {p_empirical : .4f}')
print(f'p normal : {p_normal : .4f}')
print(f'difreance : {abs(p_empirical - p_normal) : .4f}')
```

```python
smokers_df = df[df['smoker'] == 'yes']
non_smoker_df = df[df['smoker'] == 'no']

p_high_cost_given_smoker = (smokers_df['charges'] > 20000).mean()
p_high_cost_given_non_smoker = (non_smoker_df['charges'] > 20000).mean()

print(f'high cost smoker : {p_high_cost_given_smoker : .4f} (that {p_high_cost_given_smoker*100 : .2f})')
print(f'high cost non smoker : {p_high_cost_given_non_smoker : .4f} (that {p_high_cost_given_non_smoker*100 : .2f})')
```

```python
 # bayes 
prob_high_given_smoker = (df[df['smoker'] == 'yes']['charges'] > 25000).mean()
prob_smoker = (df['smoker'] == 'yes').mean()
numeator = prob_high_given_smoker * prob_smoker

prob_high =(df['charges'] > 25000).mean()

prob_smoker_given_high = numeator / prob_high

print(f'prob smoker given high :{prob_smoker_given_high} ({prob_smoker_given_high*100 : .2f}%)')
```

```python
high_cost_df = df[df['charges'] > 25000]
prob_smoker_given_high_direct = (high_cost_df['smoker'] == 'yes').mean()
print(f'prob smoker given high direct :{prob_smoker_given_high_direct} ({prob_smoker_given_high_direct*100 : .2f}%)')
```
